```python
import sqlite3
import pandas as pd

def execute_sql_query(query):
    try:
        conn = sqlite3.connect('countries.db')
        cursor = conn.cursor()

        cursor.execute(query)
        result = cursor.fetchall()

        conn.close()
        return result
    except Exception as e:
        return str(e)

def execute_python_code(code):
    try:
        result = eval(code)
        return result
    except Exception as e:
        return str(e)

def execute_csv_query(query):
    try:
        df = pd.read_csv('Cars93.csv')

        result = df.query(query)
        return result.to_dict(orient = 'records')
    except Exception as e:
        return str(e)

def q_and_a_tool(query, tool_type):
    if tool_type == "sql":
        return execute_sql_query(query)
    elif tool_type == "python_repl":
        return execute_python_code(query)
    elif tool_type == "csv":
        return execute_csv_query(query)
    else:
        return "Unknown tool type."

def create_database():
    try:
        conn = sqlite3.connect('countries.db')
        cursor = conn.cursor()

        cursor.execute('''DROP TABLE IF EXISTS countries''')
        cursor.execute('''CREATE TABLE countries (
                            id INTEGER PRIMARY KEY,
                            country_name TEXT,
                            capital TEXT,
                            population INTEGER
                        )''')

        cursor.execute("INSERT INTO countries (country_name, capital, population) VALUES ('USA', 'Washington , D.C.' , 331000000)")
        cursor.execute("INSERT INTO countries (country_name, capital, population) VALUES ('USA', 'Washington , D.C.' , 250000000)")
        cursor.execute("INSERT INTO countries (country_name, capital, population) VALUES ('Canada', 'Ottawa' , 37742154)")

        conn.commit()
        conn.close()
    except Exception as e:
        print(f"Error creating database: {e}")

create_database()
```

```python
question_1 = "SELECT * FROM countries WHERE country_name = 'USA'"
response_sql = q_and_a_tool(question_1, "sql")
print(f"Response from SQL Agent:\n{response_sql}")
```

```python
import sqlite3
from langchain_core.tools import Tool

class SQLTool:
    def __init__(self, db_path):
        self.conn = sqlite3.connect(db_path)
        self.cursor = self.conn.cursor()
        
    def query(self, query):
        try:
            self.cursor.execute(query)
            results = self.cursor.fetchall()
            if not results:
                return "No results found for the query."
            return str(results)
        except Exception as e:
            return f"Error executing query: {str(e)}"

    def close(self):
        self.conn.close()

db_path = "countries.db"
sql_tool = SQLTool(db_path)

sql_tool_function = Tool(
    name = "SQL Database Query Tool",
    func = sql_tool.query,
    description = "Execute SQL queries on the countries database."
)

query_1 = "SELECT * FROM countries WHERE population > 500000000;"

response_1 = sql_tool_function.func(query_1)

print(f"response to SQL query 1: {response_1}")
```

```python
import pandas as pd
from langchain.agents import Tool


df = pd.read_csv('Cars93.csv')
print(df.columns)


df.fillna({"Price": 0, "MPG.city": 0, "MPG.highway": 0, "EngineSize": 0}, inplace = True)

class CSVTool:
    def __init__(self, dataframe):
        self.df = dataframe

    def search(self, query):
        query = query.lower()

        if "price" in query and "of" in query:
            car_name = query.split("price of")[-1].strip()
            result = self.df[self.df['Car'].str.contains(car_name, case=False, na=False)]
            if not result.empty:
                return result[['Car', 'Price']]
            else:
                return "No car found with that name."

        elif "mpg" in query:
            if "city" in query:
                car_name = query.split("mpg city of")[-1].strip()
                result = self.df[self.df['Car'].str.contains(car_name, case = False, na = False)]
                if not result.empty:
                    return result[['Car', 'MPG.city']]
            elif "highway" in query:
                car_name = query.split("mpg highway of")[-1].strip()
                result = self.df[self.df['Car'].str.contains(car_name, case = False, na = False)]
                if not result.empty:
                    return result[['Car', 'MPG.highway']]
            else:
                return "Please specify whether you are looking for city or highway MPG."

        elif "engine size" in query:
            car_name = query.split("engine size of")[-1].strip()
            result = self.df[self.df['Car'].str.contains(car_name, case = False, na = False)]
            if not result.empty:
                return result[['Car', 'EngineSize']]
            else:
                return "No car found with that name."

        elif "type" in query:
            car_name = query.split("type of")[-1].strip()
            result = self.df[self.df['Car'].str.contains(car_name, case = False, na = False)]
            if not result.empty:
                return result[['Car', 'Type']]
            else:
                return "No car found with that name."

        else:
            return "Can you please rephrase your query to ask for a specific detail like price, engine size, or type?"

csv_tool = CSVTool(df)

csv_tool_function = Tool(
    name = "CSV Search Tool",
    func = csv_tool.search,
    description = "Search for specific car details like price, engine size, mpg and more in the Cars93 datset."
)
```

```python
from langchain_core.tools import Tool
from langchain_experimental.utilities import PythonREPL

class PythonToolWrapper:
    def __init__(self):
        self.python_repl = PythonREPL()

    def evaluate(self, query):
        try:
            result = self.python_repl.run(query)
            return result
        except Exception as e:
            return f"Error: {str(e)}"


python_tool = PythonToolWrapper()

python_tool_function = Tool(
    name = "Python Code Executor",
    func = python_tool.evaluate,
    description = "Execute Python code dynamically using REPL."
)

query_1 = "print(1+1)"
query_2 = "print('Hello, World!')"


response_1 = python_tool_function.func(query_1)
response_2 = python_tool_function.func(query_2)

print(f"Response to query 1: {response_1}")
print(f"Response to query 2: {response_2}")
```

```python
tools = [sql_tool_function, python_tool_function, csv_tool_function]

agent = initialize_agent(
    tools,
    llm = llm,
    agent_type = AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose = True
)
```

```python
def process_question(question):
    response = agent.invoke(question)
    return response

while True:
    question = input("Ask a question(type 'exit' to quit): ")
    if question.lower == "exit":
        break
    answer = process_question(question)
    print(f"Answer :{answer}")
```
